```python
from pyspark.sql.functions import col

# 1. Sample order data (simulating millions of incoming rows)
orders_data = [
    (1001, "Rahul Verma", "Kanpur", "Wireless Mouse", 2, 800),
    (1002, "Ananya Singh", "Lucknow", "Mechanical Keyboard", 1, 3500),
    (1003, "Rahul Verma", "Kanpur", "Mechanical Keyboard", 1, 3500),
    (1004, "Vikram Patel", "Varanasi", "Coffee Mug", 3, 250),
    (1005, "Rahul Verma", "Kanpur", "Wireless Mouse", 1, 800)
]

columns = ["order_id", "customer_name", "city", "product_name", "quantity", "unit_price"]

# 2. Load into a distributed Spark DataFrame
df = spark.createDataFrame(orders_data, columns)

# 3. Transform: Add a calculated 'total_amount' column using PySpark
df_transformed = df.withColumn("total_amount", col("quantity") * col("unit_price"))

# 4. Display the result cleanly
display(df_transformed)
```

```python
df_transformed.write.format("delta").mode("overwrite").saveAsTable("kanpur_orders_delta")
```

```python
%sql
SELECT
  city,
  SUM(total_amount) AS total_revenue
FROM kanpur_orders_delta
GROUP BY city
ORDER BY total_revenue DESC;
```
